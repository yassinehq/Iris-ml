# Classification des fleurs d'iris

**Objectif :** prédire l'espèce d'une fleur d'iris à partir de 4 mesures physiques.

**Données :** dataset Iris (Kaggle / UCI) : 150 fleurs, 3 espèces (50 par espèce).

**Variables d'entrée (features) :** 4 mesures en cm :
- longueur et largeur du sépale (`SepalLengthCm`, `SepalWidthCm`)
- longueur et largeur du pétale (`PetalLengthCm`, `PetalWidthCm`)

**Cible (label) :** `Species` → *Iris-setosa*, *Iris-versicolor*, *Iris-virginica*

**Type de problème :** classification supervisée multiclasse (3 classes).
